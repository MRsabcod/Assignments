```python
def StudentMarksCalculator(subjects):
    total=0

    for i in range(subjects):
        sub=int(input(f"enter your subject marks number { i+1}"))
        
        total+=sub
    avg=total/subjects
    
    return f"your total marks for {subjects} subjects is {total} and average is {avg}"
    
##ASSIGNMENT 1 PART 4
```
