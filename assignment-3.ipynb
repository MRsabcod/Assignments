```python
# part 7
accbalance=50000
while accbalance!=0:
    userSelect=input(f"1. Check Balance\n2. Deposit\n3. Withdraw\n4. Exit")
    
    if userSelect != "4":
        if userSelect=="1":
            print(f"your current balance is {accbalance}")
            continue
        elif userSelect=="2":
            deposit=int(input("enter the amount to deposit: "))
            accbalance+=deposit
            print(f"your accbalance after deposit is {accbalance}")
            continue
        elif userSelect=="3":
            withdraw=int(input("enter the amount to withdraw: "))
            accbalance-=withdraw
            print(f"your accbalance after deposit is {accbalance}")
            continue
            
    break
```

```python
# part 11
numberOfStudents=int(input())
studentsPass=0
studentsFail=0
percentagelst=[]
total=0
highestPercentage=0
lowestPercentage=0
for i in range(0,numberOfStudents):
    studentName=input("enter Name:")
    pythonMarks=float(input("your Marks in Python:"))
    mathMarks=float(input("your marks in maths:"))
    englishMarks=float(input("your marks in english:"))
    total+=pythonMarks+mathMarks+englishMarks
    percentage=(total/300)*100
    percentagelst.append(percentage)
    if percentage>=50 and pythonMarks>=40 and mathMarks>=40 and englishMarks>= 40:
        print("Pass")
        studentsPass+=1
        if percentage>= 90:
           print("A+")
        elif percentage>= 80:
           print("A")
        elif percentage>=70:
           print("B")
        elif percentage>=60:
            print("C")
        else:
            print("D") 
        
    else:
        studentsFail+=1
        print("Fail")
if  len(percentagelst) !=0:
    for i in percentagelst:
        currentval=i
        nextval=i+1
        while True:
           if currentval>nextval:
               highestPercentage=currentval
           elif currentval<nextval:
                lowestPercentage=currentval
           break

print(highestPercentage,lowestPercentage)


            
    
```
